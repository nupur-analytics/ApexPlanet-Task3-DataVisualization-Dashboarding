```python
import pandas as pd
import matplotlib.pyplot as plt
print("Task 3 Data Visualzation Started")
```

```python
data = pd.read_csv("sales (2).csv")
data.head()
```

```python
data["product_category"].value_counts().plot(kind="bar")
plt.show()
```

```python
plt.hist(data["unit_price"])
plt.show()
```

```python
plt.scatter(data["quantity"], data["total_price"])
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.bar(data["product_category"].value_counts().index,
        data["product_category"].value_counts().values)
plt.xticks(rotation=45)
plt.title("Product Category Count")
plt.subplot(1,2,2)
plt.hist(data["total_price"])
plt.title("Total Price Distribution")
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
```

```python
corr= data[["unit_price","quantity","tax","total_price"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()
```

```python
sns.pairplot(data[["unit_price","quantity","tax","total_price"]])
plt.show()
```

```python
sns.boxplot(x=data["product_category"], y=data["total_price"])
plt.xticks(rotation=45)
plt.show
```

```python
g = sns.FacetGrid(data,col="customer_type")
g.map(sns.histplot,"total_price")
plt.show()
```

```python
import plotly.express as px
```

```python
fig = px.bar(data, x="product_category", y="total_price", title="Total Price by Product Category")
fig.show()
```

```python
category_sales = data.groupby("product_category")["total_price"].sum().reset_index()
fig = px.bar(category_sales, x="product_category", y="total_price", title="Sales by Product Category")
fig.show()
```
